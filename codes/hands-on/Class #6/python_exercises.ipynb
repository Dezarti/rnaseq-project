1. Create a list m with the names of the twelve months. Subset m to create a second list containing only the first three and the last four months.


```python
# First way
m = ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December']

subset_1 = m[0:4]
subset_2 = m[8:12]

m_subset = subset_1 + subset_2

print(m_subset)
```

```python
# Second way

m = ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December']
m_subset = [m[x] for x in range(0, len(m)+1) if x<4 or 8<=x<12]
print(m_subset)
```

2. Use a for loop for printing only the elements of m that contain the letter “e”

```python
# First Way
for i in m:
  if 'e' in i:
    print(i)
```

```python
# Second way
for j in range(0, len(m)):
  if 'e' in m[j]:
    print(m[j])
```

```python
# Third one inside a list
elements_only_e = [m[j] for j in range(0, len(m)) if 'e' in m[j]]
for k in elements_only_e:
  print(k)
```

4. Define a function that accepts a list L1. The list can contain any type of elements: strings, numbers, other lists, dictionaries, etc. The function should return a new list containing only the elements of L1 that are of type string

```python
# First way
L1 = ['E1', ['E2', 'E3'], {'Key_1':1, 'Key_2':2}, 3, 4, 'Final']

def my_function(L1):
  L2 = []
  for element in L1:
    if type(element) == str:
      L2.append(element)
  return L2

result = my_function(L1)
print(result)
```

```python
# Second way
L1 = ['E1', ['E2', 'E3'], {'Key_1': 1, 'Key_2': 2}, 3, 4, 'Final']

result = [element for element in L1 if isinstance(element, str)]
print(result)
```

```python
# Third way
L1 = ['E1', ['E2', 'E3'], {'Key_1': 1, 'Key_2': 2}, 3, 4, 'Final']
result = list(filter(lambda x: isinstance(x, str), L1))
print(result)
```

5. Let's assume we want to keep track of students' grades using a Python dictionary. We have a dictionary called students where each key is the name of a students, while each value contains a nested dictionary with the grades of the student.

* Design the nested dictionaries so that you can keep track of the grades of the student. What would you use as key? What would you use as values?

* Design a function that accept the dictionary student as input and compute the average grade for each student.

* Write a small example by populating students with at least 5 students, where each student has passed at least 3 exams.

```python
# First way
students =  {'Tadeo':{'Homework':100,
                      'Projets':70,
                      'Exams': 50},
             'Camila':{'Homework':100,
                      'Projets':100,
                      'Exams': 95},
             'David':{'Homework':30,
                      'Projets':60,
                      'Exams': 10},
             'Juan':{'Homework':90,
                      'Projets':77,
                      'Exams': 74}
            }

def final_grade(students):
  grades_per_student = {}
  for key in students.keys():
    average = []
    for grade in students[key].values():
      average.append(grade)

    grades_per_student[key] = round((sum(average) / len(average)), 2)

  return grades_per_student

result = final_grade(students)
print(result)
```

```python
# Second way

students =  {'Tadeo':{'Homework':100,
                      'Projets':70,
                      'Exams': 50},
             'Camila':{'Homework':100,
                      'Projets':100,
                      'Exams': 95},
             'David':{'Homework':30,
                      'Projets':60,
                      'Exams': 10},
             'Juan':{'Homework':90,
                      'Projets':77,
                      'Exams': 74}
            }

def final_grade(students):
    return {
        name: round(sum(scores.values()) / len(scores.values()), 2)
        for name, scores in students.items()
    }

result = final_grade(students)
print(result)
```
